### Longest and Shortest Life Expectancy by Continent

For each continent, let's find:
- The country with the **longest** life expectancy
- The country with the **shortest** life expectancy

```python
# Find the country with maximum life expectancy for each continent
max_life = df.loc[df.groupby('continent')['life_expectancy'].idxmax()]

# Find the country with minimum life expectancy for each continent
min_life = df.loc[df.groupby('continent')['life_expectancy'].idxmin()]

# Display results
print("Countries with LONGEST life expectancy by continent:")
print("=" * 60)
print(max_life[['continent', 'country', 'life_expectancy']].sort_values('continent').to_string(index=False))
print("\n")
print("Countries with SHORTEST life expectancy by continent:")
print("=" * 60)
print(min_life[['continent', 'country', 'life_expectancy']].sort_values('continent').to_string(index=False))
```
