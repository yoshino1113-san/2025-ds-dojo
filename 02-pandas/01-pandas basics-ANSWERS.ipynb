Extract the top 3 countries with high GDP and long life expectancy

```python
# Extract the top 3 countries with high GDP and long life expectancy
# Aim for the top in both life_expectancy and gdp
# Here, we standardize life_expectancy and gdp, sum them, and extract the top 3 countries
%pip install scikit-learn
from sklearn.preprocessing import StandardScaler
import numpy as np

# Standardize life_expectancy and gdp
df[['life_expectancy_std', 'gdp_std']] = StandardScaler().fit_transform(df[['life_expectancy', 'gdp']])
# Sort by total score
df['score'] = df['life_expectancy_std'] + df['gdp_std']
top3 = df.sort_values('score', ascending=False).head(3)
top3[['country', 'life_expectancy', 'gdp']]
```
