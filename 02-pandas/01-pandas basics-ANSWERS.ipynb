## Life Expectancy by Continent

This analysis identifies the countries with the longest and shortest life expectancy in each continent.

```python
# Life Expectancy by Continent Analysis
# Find the countries with the longest and shortest life expectancy for each continent

from IPython.display import display, Markdown

# Get unique continents
continents = sorted(df['continent'].unique())

# Create a list to store results
results = []

# For each continent, find the countries with max and min life expectancy
for continent in continents:
    continent_df = df[df['continent'] == continent]
    
    # Get the country with longest life expectancy
    max_idx = continent_df['life_expectancy'].idxmax()
    max_country = continent_df.loc[max_idx, 'country']
    max_life = continent_df.loc[max_idx, 'life_expectancy']
    
    # Get the country with shortest life expectancy
    min_idx = continent_df['life_expectancy'].idxmin()
    min_country = continent_df.loc[min_idx, 'country']
    min_life = continent_df.loc[min_idx, 'life_expectancy']
    
    results.append({
        'Continent': continent,
        'Longest Life Expectancy': f"{max_country} ({max_life:.2f} years)",
        'Shortest Life Expectancy': f"{min_country} ({min_life:.2f} years)"
    })

# Create a DataFrame from results and display it as a markdown table
results_df = pd.DataFrame(results)
display(Markdown(results_df.to_markdown(index=False)))
```
